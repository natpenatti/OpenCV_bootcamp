### Accessing the Camera

```python
import cv2
import sys

# Mude para 0 direto, pois o sys.argv não funciona bem dentro do Jupyter
s = 0 

source = cv2.VideoCapture(s)
win_name = 'Camera Preview'
cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)

while True:
    has_frame, frame = source.read()
    if not has_frame:
        break
        
    cv2.imshow(win_name, frame)
    
    # O waitKey precisa de um tempo maior (ex: 20ms) para atualizar a imagem na tela
    # Aguarda a tecla ESC (código 27) para fechar
    if cv2.waitKey(20) & 0xFF == 27:
        break

source.release()
cv2.destroyAllWindows() 
```
